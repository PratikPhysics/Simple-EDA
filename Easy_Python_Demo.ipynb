```python
salary = [25000, 30000, 45000, 60000]

average_salary = sum(salary) / len(salary)

print("Average Salary is:", average_salary)
```

🟢 Example 2: Machine Learning in JUST 8 LINES 🤯

```python
from sklearn.linear_model import LinearRegression

experience = [[1], [2], [3], [4]]
salary = [30000, 40000, 50000, 60000]

model = LinearRegression()
model.fit(experience, salary)

print(model.predict([[5]]))
```

“This is Machine Learning — not magic, not rocket science.”

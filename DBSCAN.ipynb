```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt

# 1. Create sample data
X, _ = make_moons(n_samples=300, noise=0.05, random_state=0)

# 2. Apply DBSCAN
db = DBSCAN(eps=0.2, min_samples=5)
labels = db.fit_predict(X)

# 3. Plot the result
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='plasma')
plt.title("DBSCAN Clustering")
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()
```

```python
from sklearn.datasets import make_circles
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# 1. Generate circular data
X, _ = make_circles(n_samples=500, factor=0.5, noise=0.05)

# 2. Apply DBSCAN
db = DBSCAN(eps=0.1, min_samples=5)
labels = db.fit_predict(X)

# 3. Plot the clusters
plt.figure(figsize=(6, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='rainbow')
plt.title("DBSCAN Clustering on make_circles Data")
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()
```

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# 1. Generate blob data
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.5, random_state=0)

# 2. Apply DBSCAN
db = DBSCAN(eps=0.3, min_samples=5)
labels = db.fit_predict(X)

# 3. Plot the result
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.title("DBSCAN Clustering on make_blobs Data")
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()
```
